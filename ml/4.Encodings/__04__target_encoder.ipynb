```python
# pip install category_encoders

import pandas as pd
from category_encoders import TargetEncoder
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("/home/havin-benoy/Desktop/Pyspiders/ml/Datasets/Student_Performance.csv")
df.head()
```

```python
df.info()
```

```python
x = df.drop(columns='Performance Index')
y = df['Performance Index']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
# creating an object of target encoder

target_encoder = TargetEncoder()
```

```python
# Syntax for fit_transform():

'''
target_encoder.fit_transform(x_train[col_name(s)], y_train)
'''
```

```python
x_train['Extracurricular Activities'] = target_encoder.fit_transform(x_train['Extracurricular Activities'], y_train)

x_train
```

### Note:
- Here we need to pass both x_train and y_train
- because target encoder uses the target values to encode the x_train categorical columns
- also , here when are passing the x_train columns to encode , we can pass 1d or 2d
- so single column x_train['col_1']
- and for multiple columns x_train[['col_1','col_2','col_3',...]]
- y_train is directly passed. it is 1d

```python
# Instead of this we can follow the below approach
# It is the recommended approach

'''

target_encoder = TargetEncoder(cols = ['Extracurricular Activities'])

x_train = target_encoder.fit_transform(x_train, y_train)

x_train

'''
```

### Note :

- This the recommended method of applying the target encoding
- here while creating the object of the TargetEncoder itself we mension which all columns we need to perform the encoding by using the parameter 'cols'.
- once we create the object like that , we dont have to mention the column name in the first parameter of the fit_transform()
- we can directly call fit_transform(x_train,y_train).
- internally it will only encode the column that we mension while creating the object.
- then it fit_transform() will return the full x_train in which the column we mension is encoded.

### Note :
- consider we dont mension any column name while creating the object:

> target_encoder = TargetEncoder()
> x_train = target_encoder.fit_transform(x_train, y_train)
> x_train

- In this case , all the string based columns in the x_train will automatically be encoded.
- But this is not preffered as there might be  string based columns that are not categorical columns.
- so we select the string based categorical columns and mention it while object creation , that is the recommended approach.
