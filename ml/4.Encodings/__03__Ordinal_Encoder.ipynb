### Note :

1. categories : it helps to mention the column order and it accepts 2d
    - example:
        [['Low','Medium','High'],['High School','bachelors','masters']]
    - i.e 
    - column 1 order => ['Low','Medium','High']
    - column 2 order => ['High School','bachelors','masters']

2. handle_unknown :
    - default value is 'error' => when unseen or unknown label is present during the transform(), it will through error
    - 'use_encoded_value' => it helps to handle unknown values

3. unknown_value:
    - we use this paramter when handle_unknown = 'use_encoded_value' to assign some values to unseen labels.
    - example: 
        ord_encoder = OrdinalEncoder(
            categories= [['Low','Medium','High'],['High School','bachelors','masters']],
            handle_unknown = 'use_encoded_value',
            unknown_value = -1
        )
    - then, 
        column 1 => 'med' (This value is not present in the colummn 1 order)
        assigns => -1 for med

        column 2 => 'Associate' (this value is not present in the column 2 order)
        Assigns => -1 for 'Associate'

```python
# Actual Implementation
# Only the column person_education follows order in this dataset.
# So we will apply it only on the column 'person_education'

ord_encoder = OrdinalEncoder(
    categories= [['High School', 'Bachelor','Master', 'Associate', 'Doctorate']],
    handle_unknown= 'error',     # 'error' or 'use_encoded_value'

)
```

```python
ord_encoded_value = ord_encoder.fit_transform(x_train[['person_education']])
ord_encoded_value
```
