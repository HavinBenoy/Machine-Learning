```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("/home/havin-benoy/Desktop/Pyspiders/ml/Datasets/loan_data.csv")
df.head()
```

## Note :

- This is classification dataset
- here we have more than 1 categorical column in the input feature columns
- Our target column is not a string-based categorical column.
- As per the general rules , we usually dont applu the label encoding technique in feature column . It is applied only the target column.
- The exception is when we have only 1 categorical column in the input feature columns , then we can use the label encoder to encode that column as well.
- But in this scenario , both those cases dont exist.
- But for exploring the label encoder , we are enconding the column "person_gender" based on the label encoder, to explore how to do label encoding.

```python
df.info()
```

```python
x = df.drop(columns="loan_status")
y = df["loan_status"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
# To get the categorical values from the dataset. 
# specifically from the the input feature columns 
categorical_cols = x.select_dtypes(include = 'object')

print(categorical_cols.nunique())   
# Gives the number of unique values in each columns of the categorical columns 

print("\n\n")

display(categorical_cols.head())
```

```python
# Creating object of the LabelEncoder

label_encoder = LabelEncoder()
```

### Encoding the input feature of train data

```python

# Make sure only one categorical column is only passed as argument

encoded_values_train = label_encoder.fit_transform(x_train["person_gender"])

encoded_values_train
```

### Note:

- Instead if we try the below code , we will get the ValueError

> encoded_values_train = label_encoder.fit_transform(x_train)

- Reason:
    - Here x_train contains 13 columns which includes both numerical and categorical column.
    - Since the fit_transform() will only accept one categorical column , it will result in a ValueError'''

    - ValueError: y should be 1d array , got an array of shape (36000,13) instead

### Encoding the input feature of test data

```python
encoded_values_test = label_encoder.transform(x_test["person_gender"])

# if we pass two or more categorical columns => ValueError
# Also note that we used the transform() function and not the fit_transform()

encoded_values_test
```

```python

# Here lets create a new unseen data and use that as a future data and check what is the encoded values of it

# unseen data
new_data = pd.DataFrame({
    'person_gender':['male','female']   
})

encoded_values_unseen_data = label_encoder.transform(new_data)

display(encoded_values_unseen_data)
```

Result Analysis:
- The first unseen data is 'male' is it encoded as 1
- 'female' is encoded as 0
- why like that:
    - during the fit_transform(x_train) , the unique values of x_train (i.e 'male' and  'female') are taken
    - then they are sorted in ascending (alphabetical) order.
    - Then numbers (labels) are given starting from 1
    - since 'female' is alphabetically prior than 'male', 'female' get encoded as 0 and male gets encoded as 1
    - that is how for the column ['male','female'] we got the encoding as [1,0] 

### Note :

- A dataframe is always 2-d even if it contains 1 column , 2 column , 3 column or any number of columns. 
- here the new_data is actually 2d (since it is a DataFrame)
- But the transform() or fit_transform() of the label encoder will only accept 1d
- But it will not throw error , because eventhough the "new_data" is 2d it only have 1 column
- so it will only throw a warning. and then take that single column as the 1d input.
- But if new_data contains more than 2 columns , it will throw ValueError
