```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import (
    StandardScaler,
    RobustScaler,
    MinMaxScaler)
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("loan_data.csv")
df.head()
```

```python
x = df.drop(columns='loan_status')
y = df['loan_status']
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(
    x,
    y,
    train_size=0.8,
    random_state=42
)
```

```python
num_col = x.select_dtypes(include='number').columns
num_col
```

### There are more number of outliers for this we can use RobustScalar

```python
num_col
```

```python
num_col
```

```python
standard_scalar = StandardScaler()
robust_scalar = RobustScaler()
```

```python
r_columns = ['person_age', 'person_income', 'person_emp_exp', 'loan_amnt', 'loan_percent_income',
       'credit_score']
s_columns = ['cb_person_cred_hist_length','loan_int_rate']
```

```python
xtrain[s_columns] = standard_scalar.fit_transform(xtrain[s_columns])
xtrain[r_columns] = robust_scalar.fit_transform(xtrain[r_columns])
xtrain.head()
```

```python
xtrain['cb_person_cred_hist_length'].mean()
```

```python
xtrain[s_columns].std()
```

```python
xtest[s_columns] = standard_scalar.transform(xtest[s_columns])
xtest[r_columns] = robust_scalar.transform(xtest[r_columns])
xtest.head()
```
