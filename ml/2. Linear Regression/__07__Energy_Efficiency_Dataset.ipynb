```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_absolute_error, root_mean_squared_error,mean_squared_error
```

```python
df = pd.read_csv("/home/havin-benoy/Desktop/Pyspiders/ml/Datasets/ENB2012_data.csv")

display(df.head())
```

```python
# Giving appropriate naming as per the official documentation in the kaggle website for this dataset

name_changes = {
"X1":"Relative Compactness",
"X2":"Surface Area",
"X3":"Wall Area",
"X4":"Roof Area",
"X5":"Overall Height",
"X6":"Orientation",
"X7":"Glazing Area",
"X8":"Glazing Area Distribution",
"Y1":"Heating Load",
"Y2":"Cooling Load"
}

df = df.rename(columns= name_changes)

df.head()
```

```python
display(df.corr())
```

Note :
- Here we have two output columns.
- We are choosing the "Cooling Load" as the target column
Also, 
- Here we wanna explore the linear regression in the non linear data .
- so we are only choosing the features that are having low correlation with the output column
- i.e we are choosing only the below columns only as the independent columns:
    1. Orientation
    2. Glazing Area Distribution

```python
df = df[["Orientation","Glazing Area Distribution","Cooling Load"]]

display(df.corr())
```

conclusion from the correlation matrix:
- Now this dataset only contains the features that are having low correlation.
- hence the data is non linear
- we cannot use linear regression to model the relationship between the features of this dataframe
- But for the experimentation purpose, we are proceeding with the linear regression.


```python

x = df.drop(columns = "Cooling Load")
y = df["Cooling Load"]

x_train,x_test,y_train,y_test = train_test_split(x,y,train_size=0.8,random_state=42)

model_lr = LinearRegression()

model_lr.fit(x_train,y_train)
```

```python

model_lr.intercept_
```

```python

# Slope

model_lr.coef_
```

```python
# Predictions

y_train_predicted = model_lr.predict(x_train)

y_test_predicted = model_lr.predict(x_test)
```

# Regression Evaluation metrics

1. Mean Squared Error (MSE)
2. Root Mean Squared Error (RMSE)
3. Mean Absolute Error (MAE)
4. r^2 score

```python
# 4. r2 score

train_r2_score =r2_score(y_train,y_train_predicted)

test_r2_score= r2_score(y_test,y_test_predicted)


print(train_r2_score, test_r2_score)
```

```python
# 2. Root Mean Squared Error (RMSE)

train_rmse =root_mean_squared_error(y_train,y_train_predicted)

test_rmse = root_mean_squared_error(y_test,y_test_predicted)


print(train_rmse, test_rmse)
```

```python
# 3. Mean Absolute Error (MAE)

train_mae =mean_absolute_error(y_train,y_train_predicted)

test_mae = mean_absolute_error(y_test,y_test_predicted)


print(train_mae, test_mae)
```

```python
# Error percentage based on the MAE .

# To check if an mae value is high or not we should use it to calculate the error percentage
#  error percentage based on mae equation is : (average(y_train) / mae )* 100

avg_of_y_train = sum(y_train)/len(y_train)

print("Average of Y train: ",avg_of_y_train)

error_percentage_based_on_mae = (avg_of_y_train/train_mae)*100

print("Error percentage based on mae: ",error_percentage_based_on_mae)
```

```python
# 1. Mean Squared Error (MSE)

train_mse = mean_squared_error(y_train,y_train_predicted)

test_mse = mean_squared_error(y_test,y_test_predicted)


print(train_mse, test_mse)
```

## Final conclusion:
- r2_score is very low for both train and test splits of the data.
- That means the model didnt understood the internal pattern or is not capable of understanding the complex non linear pattern in the data.
- Since the test r2 score is low , model will not perform well in the future unseen data as well.
