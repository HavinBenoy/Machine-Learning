```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_absolute_error, root_mean_squared_error,mean_squared_error
```

```python
df = pd.read_csv("/home/havin-benoy/Desktop/Pyspiders/ml/Datasets/auto-mpg.csv")

display(df.head())
```

```python
df["car name"].unique()
```

```python
# Since the "car name" is like a identifier kind of column, 
# we are dropping it from the df for the linear model training


df = df.drop(columns = "car name")

df.head()
```

```python
df.info()
```

```python
# Here the "horsepower" column is having string datatype .
# Hence we are inspecting that column and then converting it into int values

print(df["horsepower"].unique())
```

```python
# We could see '?' on the list

print(df["horsepower"].value_counts()['?'])
```

```python
# Checking if the other column values of the rows having '?' on the "horsepower" is valid or null or '?'
df[df["horsepower"]=='?']
```

### Note

- so there are 6 rows in which the horsepower value is '?'.

- Also ,the other columns of those rows are valid data.

- so removing those rows blindly is not a good approach

- so , lets find the mean of the rest of the values of the colunmns and fill that on these positions of '?' on "hoursepower" column.

```python
# Findinge Mean
mean_of_horsepower = int(df[df["horsepower"]!='?']["horsepower"].astype(int).mean())
print(f"Mean : {mean_of_horsepower}")
df["horsepower"] = df["horsepower"].replace({"?":mean_of_horsepower})
df["horsepower"] = df["horsepower"].astype(int)
df["horsepower"].unique()
```

```python
df.info()
```

```python
display(df.corr())
```

conclusion from the correlation matrix :
- here "mpg" is the target column
- here , columns like "displacement", "weight", ...are having high correlation with the target feature.
- Hence this data is linear and we can use linear regression to model the relationship between the data.

```python
# x = df[["Temperature (°C)","Ice Cream Sales (units)"]]
x = df.drop(columns = "mpg")
y = df["mpg"]

x_train,x_test,y_train,y_test = train_test_split(x,y,train_size=0.8,random_state=42)

model_lr = LinearRegression()

model_lr.fit(x_train,y_train)
```

```python

model_lr.intercept_
```

```python

# Slope

model_lr.coef_
```

```python
# Predictions

y_train_predicted = model_lr.predict(x_train)

y_test_predicted = model_lr.predict(x_test)
```

# Regression Evaluation metrics

1. Mean Squared Error (MSE)
2. Root Mean Squared Error (RMSE)
3. Mean Absolute Error (MAE)
4. r^2 score


```python
# 4. r2 score

train_r2_score =r2_score(y_train,y_train_predicted)

test_r2_score= r2_score(y_test,y_test_predicted)


print(train_r2_score, test_r2_score)
```

```python
# 2. Root Mean Squared Error (RMSE)

train_rmse =root_mean_squared_error(y_train,y_train_predicted)

test_rmse = root_mean_squared_error(y_test,y_test_predicted)


print(train_rmse, test_rmse)
```

```python
# 3. Mean Absolute Error (MAE)

train_mae =mean_absolute_error(y_train,y_train_predicted)

test_mae = mean_absolute_error(y_test,y_test_predicted)


print(train_mae, test_mae)
```

```python
# Error percentage based on the MAE .

# To check if an mae value is high or not we should use it to calculate the error percentage
#  error percentage based on mae equation is : (average(y_train) / mae )* 100

avg_of_y_train = sum(y_train)/len(y_train)

print("Average of Y train: ",avg_of_y_train)

error_percentage_based_on_mae = (avg_of_y_train/train_mae)*100

print("Error percentage based on mae: ",error_percentage_based_on_mae)
```

```python
# 1. Mean Squared Error (MSE)

train_mse = mean_squared_error(y_train,y_train_predicted)

test_mse = mean_squared_error(y_test,y_test_predicted)


print(train_mse, test_mse)
```

## Final conclusion:

- r2_score is  high for both train and test splits.
- That means the model understood the internal pattern in the data and will perform well on the unseen future data as well.
- The reason for this that the data is linear data and the linear regression model was able to capture linear data pattern.
